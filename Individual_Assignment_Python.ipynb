#### 2.2.1 Load and Verify the Happiness Dataset
This section shows the process of loading, verifying, and preparing the 2019 World Happiness dataset (https://s3.amazonaws.com/happiness-report/2019/Chapter2OnlineData.xls). The dataset is retrieved from GitHub and loaded into a pandas DataFrame. Essential steps include ensuring the dataset is in the correct format, verifying its structure, and renaming columns for clarity.

The Happiness dataset measures and ranks countries by their population's overall happiness, considering the following factors:

- **Overall rank**: Rank of countries from highest to lowest happiness scores.
- **Country or region:** Name of the country or region surveyed.
- **Score:** Average happiness score from 0 (worst) to 10 (best), based on the Cantril Ladder.
- **GDP per capita:** Contribution of economic output (adjusted for purchasing power) to happiness; higher values indicate greater influence of prosperity.
- **Social support:** Strength of community networks, measured by respondents' ability to count on others in need.
- **Healthy life expectancy:**  Impact of life expectancy on happiness, normalized into a contribution score.
- **Freedom in life choices:** Perceived freedom in decision-making; higher values suggest more autonomy.
- **Generosity:** Altruism level based on responses about donations and helping others.
- **Perceptions of corruption:** Effect of perceived government and business corruption on happiness; lower values signify less corruption and higher happiness.

## 2.3 Merging the DataFrames
This step combines the datasets (Happiness, GDP per Capita, Mental Illness Prevalence, and CO2 Emissions) into a single DataFrame. To ensure a successful merge, column names are standardised across the datasets, and country names are aligned using a mapping dictionary (https://pandas.pydata.org/docs/reference/api/pandas.Series.map.html).

Initially, an inner join was used to retain only rows with data available across all datasets. However, this resulted in significant data loss. Therefore, a left join is used instead to retain as many rows as possible (Mckinney, 2017).


````python
# Rename columns to ensure consistency across datasets
data_mentalillness_2019 = data_mentalillness_2019.rename(columns={"Entity": "Country or region"})  # Rename 'Entity' in mental illness data
data_co2_2019 = data_co2_2019.rename(columns={"Country": "Country or region"})  # Rename 'Country' in CO2 data
data_gdp_2019 = data_gdp_2019.rename(columns={"Country Name": "Country or region"})  # Rename 'Country' in CO2 data


# Standardise country names using a mapping dictionary
country_mapping = {
    'United States of America': 'United States',
    'South Korea': 'Republic of Korea',
    'Trinidad and Tobago': 'Trinidad & Tobago',
    'Congo': 'Congo (Brazzaville)',
    'Congo (Kinshasa)': 'Congo',
    'Viet Nam': 'Vietnam',
    "Cote d'Ivoire": "Ivory Coast",
    "Côte d'Ivoire": "Ivory Coast",
    'Occupied Palestinian Territory': 'Palestinian Territories',
    'Eswatini': 'Swaziland',
    'Czechia': 'Czech Republic',
    'Palestine': 'Palestinian Territories',
    'Northern Cyprus': 'Cyprus',
    'Congo, Dem. Rep.': 'Congo',
    'Congo, Rep.': 'Congo',
    'Iran, Islamic Rep.': 'Iran',
    'Russian Federation': 'Russia',
    'Egypt, Arab Rep.': 'Egypt',
    'Gambia, The': 'Gambia',
    'West Bank and Gaza': 'Palestinian Territories',
    'Venezuela, RB': 'Venezuela',
    'Hong Kong SAR, China': 'Hong Kong',
    'Kyrgyz Republic': 'Kyrgyzstan',
    'Lao PDR': 'Laos',
    'Korea, Rep.': 'Republic of Korea',
    'Slovak Republic': 'Slovakia',
    'Syrian Arab Republic': 'Syria',
    'Turks and Caicos Islands': 'Turkey',
    'Yemen, Rep.': 'Yemen',
    'Taiwan': 'Taiwan',
    'South Sudan': 'South Sudan'
}

# Apply the mapping to standardise country names across all datasets
data_happiness['Country or region'] = data_happiness['Country or region'].replace(country_mapping)
data_mentalillness_2019['Country or region'] = data_mentalillness_2019['Country or region'].replace(country_mapping)
data_co2_2019['Country or region'] = data_co2_2019['Country or region'].replace(country_mapping)
data_gdp_2019['Country or region'] = data_gdp_2019['Country or region'].replace(country_mapping)


# Merge datasets in two steps:

# A) Merge Happiness dataset with the Mental Illness dataset
first_merged_data = pd.merge(
    data_happiness,
    data_gdp_2019,
    on='Country or region',
    how='left'
)

# B) Merge Happiness dataset with the Mental Illness dataset
merged_data = pd.merge(
    first_merged_data,
    data_mentalillness_2019,
    on='Country or region',
    how='left'
)

# C) Merge the result with the CO2 dataset
merged_data = pd.merge(
    merged_data,
    data_co2_2019,
    on='Country or region',
    how='left'
)
````

#### 3.1 Data Validation and Cleaning

This section involves validating and cleaning the merged dataset to ensure consistency and readiness for analysis. Key tasks include:
- Identifying negative values in numerical columns (e.g., where negative values are not logical).
- Checking for entirely missing columns or rows.
- Validating categorical data (e.g., invalid country names or unusual entries).
- Summarising all identified issues in a report for further action.


````python
def validate_and_clean_data(df):
    # Validate and clean the DataFrame for inconsistent entries.    
    # Initialise a report dictionary to record issues
    report = {}
    
    # Go through each column in the DataFrame
    for column in df.columns:
        dtype = df[column].dtype
        print(f"Validating column: {column} (dtype: {dtype})")
        
        if dtype in ['float64', 'int64']:
            # Check for negative values in numeric columns
            negative_values = df[column][df[column] < 0]
            if not negative_values.empty:
                print(f"  - Found {len(negative_values)} negative values in {column}.")
                report[column] = "Contains negative values"
            
            # Check if the column is entirely NaN
            if df[column].isnull().all():
                print(f"  - Column {column} is entirely NaN!")
                report[column] = "Entirely NaN column"
        
        elif dtype == 'object' or str(dtype) == 'category':
            # Analyse categorical data: Check for unique values
            unique_values = df[column].unique()
            print(f"  - {len(unique_values)} unique values in {column}.")
            
            # Example: Detect invalid entries in the "Country or region" column
            if column == "Country or region":
                invalid_countries = df[column][~df[column].str.isalpha()]
                if not invalid_countries.empty:
                    print(f"  - Found potential invalid country names: {invalid_countries.unique()}")
                    report[column] = "Potential invalid entries"
        
        # Check for missing values in the column
        missing_count = df[column].isnull().sum()
        if missing_count > 0:
            print(f"  - {missing_count} missing values in {column}.")
            report[column] = "Contains missing values"
    
    # Output a summary report
    print("\nValidation and cleaning report:")
    for col, issue in report.items():
        print(f"  - {col}: {issue}")
    
    # Return the original DataFrame and the report
    return df, report

validated_data, validation_report = validate_and_clean_data(merged_data)

# Inspecting the merged DataFrame for issues
#print("\nInitial Report:", validation_report)

"""
The initial validation flagged names with spaces or special characters (e.g., "United States" or "Côte d’Ivoire") as invalid. 
To address this, the validation is refined to allow spaces, hyphens, and apostrophes in country names, as these are valid characters. 
The report generated lists any truly invalid entries for review and correction.
"""

# Function to validate country names with refined rules
def validate_country_names(df, column="Country or region"):
    """
    Validate entries in the specified country column by checking for valid characters.

    Valid characters:
    - Alphabetical characters (a-z, A-Z)
    - Spaces (' ')
    - Hyphens ('-')
    - Apostrophes ("'")
    - Ampersands ('&')

    """
    # Regular expression to allow valid country names
    invalid_countries = df[column][~df[column].str.match(r"^[a-zA-Z\s\-\&']+$", na=False)].unique()

    # Generate a validation report
    report = {
        "Invalid Country Names": list(invalid_countries) if invalid_countries.size > 0 else "No issues found"
    }
    return report

# Validate country names in the merged dataset
country_report = validate_country_names(merged_data)

# Display the validation report
print("\nCountry Validation Report:", country_report)
````

To better understand the distribution of wealth across countries, a new column was created to classify GDP per capita into four income groups:

- Low Income

These classifications were determined based on percentiles:
- 25th Percentile: Low-income threshold
- 50th Percentile: Middle-income threshold
- 75th Percentile: High-income threshold

The classifications allow for easier comparisons and visual representation of economic differences and disparities.

````python
# Calculate percentiles for GDP per capita
low_threshold = merged_data['GDP per capita'].quantile(0.25)
middle_threshold = merged_data['GDP per capita'].quantile(0.5)
high_threshold = merged_data['GDP per capita'].quantile(0.75)

# Function to classify GDP per capita into income groups based on percentiles
def classify_gdp_percentile(gdp):
    if gdp > high_threshold:
        return 'High Income'
    elif gdp > middle_threshold:
        return 'Upper-Middle Income'
    elif gdp > low_threshold:
        return 'Lower-Middle Income'
    else:
        return 'Low Income'

# Apply the classification
merged_data['Income Group'] = merged_data['GDP per capita'].apply(classify_gdp_percentile)

# Visualising GDP per Capita Distribution with Thresholds
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(merged_data['GDP per capita'], bins=20, color='#440154', edgecolor='black', alpha=0.7, label='GDP per Capita')
plt.axvline(low_threshold, color='red', linestyle='dashed', linewidth=1, label='Low Income Threshold (25th)')
plt.axvline(middle_threshold, color='orange', linestyle='dashed', linewidth=1, label='Middle Income Threshold (50th)')
plt.axvline(high_threshold, color='green', linestyle='dashed', linewidth=1, label='High Income Threshold (75th)')
plt.title('GDP per Capita Distribution with Income Group Thresholds')
plt.xlabel('GDP per Capita')
plt.ylabel('Frequency')
plt.legend()
plt.show()

# Ensure the distribution makes sense
merged_data[['Country or region', 'GDP per capita', 'Income Group']].groupby('Income Group').describe()
````

````python
# Get the top 10 countries by Happiness Score
top_10_countries = merged_data.nlargest(10, "Happiness Score")

"""
At first, I had used Mental Helath Burden as one of the metrics to look at. However, since all of the other metrics were positive
where a higher value was better, I thought that it would not make sense to have a metric where a lower value was better.
Hence, I decided to slightly adjust Mental Health Burden by flipping it around to make Positive Mental Health.
"""

# Create a new metric: Positive Mental Health
top_10_countries["Positive Mental Health"] = (
    top_10_countries["Mental Health Burden"].max() - top_10_countries["Mental Health Burden"]
)

# Columns to normalise for the visualisation
columns_to_aggregate = [
    "Happiness Score",
    "Positive Mental Health",  
    "Social Resilience",
    "Happiness Efficiency",
    "Sustainable Happiness Index",
    "Resilient Happiness"
]

"""
In order to improve comparisons between the different metrics by making them more fair, they were normalised using 
min-max scaling.

Originally, the country with the lowest values would not show anything on the graph, which I though could be misleading, so I added an 
offset to make sure every country was visible on the graphs.
"""

# Apply Min-Max Scaling with an offset
top_10_countries_norm = top_10_countries.copy()
epsilon = 0.01  # Small positive offset
for col in columns_to_aggregate:
    min_val = top_10_countries[col].min()
    max_val = top_10_countries[col].max()
    top_10_countries_norm[col] = ((top_10_countries[col] - min_val) / (max_val - min_val)) * (1 - epsilon) + epsilon

# Sort countries by normalised Happiness Score
sorted_countries = top_10_countries_norm.sort_values("Happiness Score", ascending=True)

# Calculate correlation matrix using normalised data
correlation_matrix = top_10_countries_norm[columns_to_aggregate].corr()
````

## 5.5 Data Analysis for Mental Health Burden and CO2 Emissions Across Happiness Levels and Income Groups

**Objective**
To analyse the relationship between happiness levels, income groups, mental health burden, and CO2 emissions. Log-transforming CO2 emissions provides a clearer comparison by reducing the impact of extreme values, and boxplots help visualise the distribution and variability within groups.

**Justifications for Methods**
1. **Log Transformation**: This method is applied to `Total CO2 Emissions` to normalise skewed data and make it more interpretable for analysis.
2. **Boxplots**: Displays the distribution of values within groups (Happiness Level and Income Group), highlighting medians, variability, and outliers.
3. **Annotating Outliers**: Annotating outliers to bring attention to extreme cases.

**Steps**
1. Log-transform `Total CO2 Emissions` to create `Log CO2 Emissions`.
2. Define a function to annotate outliers based on interquartile ranges (IQR).

````python
# Log-transform the Total CO2 Emissions
merged_data['Log CO2 Emissions'] = np.log1p(merged_data['Total CO2 Emissions'])  # log(1+x) avoids log(0) issues

# Define a function for annotating outliers
def annotate_outliers(ax, data, group_col, metric, color, alignment='bottom'):
    unique_groups = data[group_col].unique()
    for group in unique_groups:
        Q1 = data[data[group_col] == group][metric].quantile(0.25)
        Q3 = data[data[group_col] == group][metric].quantile(0.75)
        IQR = Q3 - Q1
        lower_threshold = Q1 - 1.5 * IQR
        upper_threshold = Q3 + 1.5 * IQR
        
        outliers = data[
            (data[group_col] == group) & 
            ((data[metric] > upper_threshold) | (data[metric] < lower_threshold))
        ]
        
        for i, row in outliers.iterrows():
            ax.text(
                x=list(unique_groups).index(group), 
                y=row[metric], 
                s=row['Country or region'], 
                color=color, 
                fontsize=9, 
                ha='right' if i % 2 == 0 else 'left',
                va=alignment if row[metric] > upper_threshold else 'top'
            )
````

# 8. Conclusion

This project provided meaningful insights into the complex relationships between happiness, wealth, mental health, and environmental factors to successfully address the main objectives set out at the start. One surprising finding was that happier and wealthier countries reported higher rates of mental health prevalence. While this initially seemed counterintuitive, it likely reflects reduced stigma in these nations, enabling more people to seek mental health diagnoses. This suggests that happiness and higher income levels do not equate to fewer mental health challenges but rather indicate the presence ofbetter systems for recognising and addressing these issues (Thornicroft, 2008).

The relationship between CO2 emissions, happiness, and wealth was seen to be more complicated. Wealthy, happy countries like the United States and Qatar exhibit high emissions due to industrialised economies and high-consumption lifestyles (United Nations Environment Programme, 2019). Similarly, developing nations such as China and India show high emissions due to industrialisation and large populations. In contrast, countries like Finland and Denmark demonstrate that a strong sustainability focus can reduce emissions while maintaining wealth and happiness (OECD, 2020). High emissions can negatively impact happiness by harming the environment and causing health problems (United Nations Environment Programme, 2019). This highlights the need for policies that balance growth with environmental and societal well-being and happiness.

This study also identified patterns distinguishing the happiest countries. Countries like Finland and Denmark excel across multiple metrics, including sustainability, social resilience, and mental health, achieving a balanced approach to happiness (Helliwell et al., 2020). Others, like Iceland, achieve high Happiness Scores by excelling in specific areas such as sustainability, driving their overall Happiness Score (Government of Iceland, 2019). These findings suggest that the happiest countries prioritise long-term well-being and environmental sustainability, despite their different approaches.

The results have significant implications. Governments should prioritise investments in mental health services, even in wealthier countries, by creating accessible programs and reducing stigma through public awareness campaigns. Addressing mental health remains essential even in countries with higher happiness levels. For businesses, particularly in wealthier countries, the link between CO2 emissions and happiness emphasises the importance of adopting sustainable practices (Elkington, 1997). Companies can reduce their carbon footprints through energy-efficient operations while maintaining employee well-being and customer satisfaction in order to align with public expectations and enhance brand reputation.

However, the analysis has limitations. The data does not fully explain the causes behind observed relationships, such as whether wealth and happiness lead to better mental health systems or whether cultural norms influence such outcomes. Additionally, the focus on total CO2 emissions overlooks per capita emissions and other sustainability indicators, such as renewable energy usage. Including such variables in future research could offer a more comprehensive understanding of the environmental impact of happiness. Longitudinal data would further enhance the analysis, tracking how happiness, mental health, and emissions evolve over time and respond to global events like economic crises or climate policies.

In conclusion, this project revealed imactful and sometimes unexpected relationships between happiness, wealth, mental health, and CO2 emissions. The findings gave insights about wealth and happiness, revealing the importance of strong mental health systems and sustainable practices in improving quality of life. By addressing these interconnected factors, policymakers and businesses can work together to create a future where prosperity, happiness, and environmental responsibility can advance together.
